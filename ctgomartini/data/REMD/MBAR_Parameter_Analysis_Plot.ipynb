```python
import matplotlib.pyplot as plt
import pickle
import numpy as np
plt.style.use('adjust_ys3')

with open('AnalysisParameter/selected_state_results.pkl', 'rb') as f:
    selected_state_results = pickle.load(f)

ref_dist_closed = 34.2745
ref_dist_open = 48.7876

# Create a figure with two subplots side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))

# First plot - Free Energy Differences
colors = plt.cm.tab20(np.linspace(0, 1, len(selected_state_results)))
for (key, results), color in zip(selected_state_results.items(), colors):
    # if key not in [4,5,6]: continue
    ax1.plot(results["cv_values"], 
             results["pmf"]-results['metrics']["basin1_pmf"], 
             label=f'state={key}',
             color=color)
    ax1.fill_between(results["cv_values"], 
                    results["pmf"] - results['metrics']["basin1_pmf"] - results["pmf_uncertainty"], 
                    results["pmf"] - results['metrics']["basin1_pmf"] + results["pmf_uncertainty"], 
                    alpha=0.2,
                    color=color)
# ax1.axvline(ref_dist_closed, color='black', ls='dashed', lw=2)
# ax1.axvline(ref_dist_open, color='black', ls='dashed', lw=2)
ax1.set_xlabel("dRMS (Å)")
ax1.set_ylabel("Free Energy (kJ/mol)")
ax1.legend(loc='upper right', fontsize=20)

# Second plot - Keq vs start ratio
key_list = []
keq_list = []
for key, value in selected_state_results.items():
    ax2.scatter(key, np.log(value['metrics']['keq']), color='red')
    key_list.append(key)
    keq_list.append(value['metrics']['keq'])
ax2.plot(key_list, np.log(keq_list), color='black')
ax2.set_xlabel("State")
ax2.set_ylabel("Log(Keq)")
# ax2.set_ylim(-10, 10)

plt.tight_layout()  # Adjust spacing between subplots
plt.show()
```
